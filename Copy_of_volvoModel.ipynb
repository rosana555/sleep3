```python
## Densenet setup??

%cd /content/Volvo-DataX

# 1. COMPLETELY reset TensorFlow state
tf.keras.backend.clear_session()
tf.config.run_functions_eagerly(True)  # Force eager execution globally

# 2. Load model with EAGER EXECUTION FORCED
def load_densenet_model():
    # Try loading with legacy format
    try:
        model = tf.keras.models.load_model(
            'densenet_2.hdf5',
            compile=False
        )
        model.compile(run_eagerly=True)  # Force eager execution
        return model
    except:
        pass

    # Fallback to JSON loading
    with open('densenet_model.json', 'r') as f:
        model = tf.keras.models.model_from_json(
            f.read(),
            custom_objects={'Model': tf.keras.Model}
        )
    model.load_weights('densenet_2.hdf5')
    model.compile(run_eagerly=True)  # Critical for eager execution
    return model

model_j = load_densenet_model()

# 3. SAFE prediction function
def pred_func(X_test):
    """Bulletproof prediction with eager enforcement"""
    # Ensure proper input shape
    X_test = np.expand_dims(X_test, 0) if X_test.ndim == 4 else X_test

    # Convert to float32 and normalize
    X_test = X_test.astype(np.float32) / 255.0

    # Use legacy prediction method
    with tf.device('/CPU:0'):  # Force CPU execution if GPU issues persist
        predictions = model_j(X_test, training=False)  # Direct call
        return np.argmax(predictions[0].numpy(), axis=0)

# Test prediction
test_input = np.random.rand(1, 100, 100, 16, 3).astype(np.float32)
print("Test prediction:", pred_func(test_input))
```

```python
# Yolo setup - testing
from ultralytics import YOLO

# Load YOLOv8 model and move to CPU
yolo = YOLO('yolov8n.pt')
yolo.to('cpu')


def detect_humans(frame: np.ndarray, conf_thresh=0.5) -> np.ndarray:
    """
    Runs YOLOv8 inference on a BGR frame and returns detections of humans only.

    Args:
        frame: Input image in BGR format.
        conf_thresh: Minimum confidence threshold for detections.

    Returns:
        A NumPy array of shape (M, 5): [x1, y1, x2, y2, confidence]
    """
    # Convert BGR to RGB and run inference
    results = yolo(frame[..., ::-1])

    # Extract raw detections
    boxes = results[0].boxes
    if boxes is None or boxes.shape[0] == 0:
        return np.zeros((0, 5), dtype=float)

    # Retrieve coordinates, confidence scores, and class IDs
    xyxy = boxes.xyxy.cpu().numpy()       # shape (N,4)
    conf = boxes.conf.cpu().numpy().reshape(-1, 1)  # shape (N,1)
    classes = boxes.cls.cpu().numpy().astype(int)   # shape (N,)

    # COCO class ID for person is 0; keep only human detections
    mask = (classes == 0) & (conf.flatten() >= conf_thresh)
    if not mask.any():
        return np.zeros((0, 5), dtype=float)

    # Stack filtered boxes and confidences
    detections = np.hstack((xyxy[mask], conf[mask]))
    return detections
```

```python

def run_model():
    """
    Main processing loop:
    1) Change to project root and open video
    2) Set up video writer if output flag provided
    3) Detect, track, and estimate pose for each human/cyclist
    4) Predict intent and annotate frames
    5) Save output video
    """
    print('Processing started.......')
    start_time = time.time()

    # 1) Set project root and change working directory
    project_root = '/content/Volvo-DataX'
    os.chdir(project_root)
    print("Switched to project root:", os.getcwd())

    # 2) Open input video
    video_path = os.path.join(project_root, 'data', 'JAAD_test_video_0339.mp4')
    print("Opening video at:", video_path)
    vid = cv2.VideoCapture(video_path)
    if not vid.isOpened():
        raise RuntimeError(f"Cannot open video {video_path!r}")

    # 3) Initialize output writer if needed
    out = None
    if FLAGS.output:
        w = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
        h = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))
        fps = int(vid.get(cv2.CAP_PROP_FPS)) or 30
        fourcc = cv2.VideoWriter_fourcc(*FLAGS.output_format)
        output_path = os.path.join(project_root, FLAGS.output)
        out = cv2.VideoWriter(output_path, fourcc, fps, (w, h))
        print(f"Writing output video to: {output_path} @ {fps} FPS, {w}x{h}")

    # 4) Initialize tracker and data buffer
    mot_tracker = Sort()
    rolling_data = {}
    frame_idx = 0

    # 5) Frame-by-frame processing
    while True:
        ret, img = vid.read()
        if not ret:
            break
        frame_idx += 1
        img_orig = img.copy()

        # Detect humans and cyclists in the frame
        detections = detect_humans_and_cyclists(
            img_orig,
            conf_thresh=FLAGS.yolo_score_threshold
        )
        trackers = mot_tracker.update(detections)

        # Loop over each tracked object
        for x1, y1, x2, y2, track_id in trackers.astype(int):
            # Compute bounding box dimensions
            w_box, h_box = x2 - x1, y2 - y1
            if w_box <= 0 or h_box <= 0:
                continue  # skip invalid boxes

            # Extract and validate crop region
            crop = img_orig[y1:y2, x1:x2]
            if crop.size == 0:
                continue

            # ----- Pose estimation -----
            try:
                # Resize crop to pose model's expected input
                crop_resized = cv2.resize(crop, (432, 368))
            except cv2.error as e:
                print(f"Skipping pose resize for track {track_id}: {e}")
                continue

            # Run pose estimator
            humans = pose_model.inference(
                crop_resized,
                resize_to_default=False,
                upsample_size=resize_out_ratio
            )
            humans.sort(key=lambda h: h.score, reverse=True)
            skeleton = TfPoseEstimator.draw_humans(crop_resized, humans, imgcopy=True)

            # Resize skeleton to original bbox size
            # Ensure exact match to slice dimensions
            target_slice = img_orig[y1:y2, x1:x2]
            th, tw = target_slice.shape[:2]
            try:
                skeleton_resized = cv2.resize(skeleton, (tw, th))
                img_orig[y1:y2, x1:x2] = skeleton_resized
            except cv2.error as e:
                print(f"Skipping skeleton paste for track {track_id}: {e}")

            # ----- Intent prediction and annotation -----
            seq = rolling_data.get(track_id, [])
            if len(seq) < 16:
                pad = seq[-1] if seq else cv2.resize(crop, (100,100))
                seq = seq + [pad] * (16 - len(seq))
            seq_array = np.stack(seq, axis=2)[None, ...]
            intent = pred_func(seq_array)

            # Draw bounding box and labels
            color = (0,0,255) if intent == 1 else (0,255,0)
            cv2.rectangle(img_orig, (x1, y1), (x2, y2), color, 2)
            cv2.putText(
                img_orig,
                str(track_id),
                (x1, y1-5),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                color,
                2
            )
            cv2.putText(
                img_orig,
                f"Frame: {frame_idx}",
                (10, 30),
                cv2.FONT_HERSHEY_COMPLEX_SMALL,
                1,
                (255, 0, 0),
                2
            )

            # ----- Update rolling buffer for intent model -----
            crop_for_buf = img_orig[y1:y2, x1:x2]
            if crop_for_buf.size == 0:
                crop_buffer = np.zeros((100, 100, 3), np.uint8)
            else:
                try:
                    crop_buffer = cv2.resize(crop_for_buf, (100, 100))
                except cv2.error:
                    crop_buffer = np.zeros((100, 100, 3), np.uint8)

            buf = rolling_data.get(track_id, [])
            if len(buf) >= 16:
                buf.pop(0)
            buf.append(crop_buffer)
            rolling_data[track_id] = buf

        # Write annotated frame to output video
        if out:
            out.write(img_orig)

    # Cleanup resources
    vid.release()
    if out:
        out.release()
    cv2.destroyAllWindows()

    elapsed = time.time() - start_time
    print(f'Processing completed in {elapsed:.2f} seconds.')
    if FLAGS.output:
        print(f'Output video saved to: {output_path}')


print("Done!")
```
